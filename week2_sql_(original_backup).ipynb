```python
import pandas as pd, sqlite3
df = pd.read_csv("https://raw.githubusercontent.com/SujalS14/Daily-News-for-Stock-Market-Prediction_BAAI_CP/refs/heads/main/Combined_News_DJIA%202.csv")
df = df.drop_duplicates().dropna()
df.columns = (df.columns.str.strip().str.lower().str.replace(" ", "_"))
```

```python
conn = sqlite3.connect(":memory:")
df.to_sql("data", conn, index=False)
def q(sql):return pd.read_sql(sql,conn)
```

My table has 1986 rows.

One Row = One day

Numbers to add up: label

Groups to Compare: year, month

```python
q("""
SELECT COUNT(*) AS down_days
FROM data
WHERE label = 0
""")
```

```python
q("""
SELECT label, COUNT(*) AS days,
  SUM(1) AS total
FROM data
GROUP BY label
""")
```

```python
q("""
SELECT date,
  SUM(label) AS total
FROM data
GROUP BY date
ORDER BY total DESC
LIMIT 2
""")
```

Q1. How many days had a label of 1 (market up)?”

```python
q("""
SELECT COUNT(*) AS up_days
FROM data
WHERE label = 1
""")
```

Q2. Which year had the most days with a label of 1?”

```python
q("""
SELECT SUBSTR(date, 1, 4) AS year, SUM(label) AS up_days
FROM data
GROUP BY year
ORDER BY up_days DESC
LIMIT 2
""")
```

Q3. What are the top 5 months in 2008 with the most down days?

```python
q("""
SELECT SUBSTR(date, 1, 7) AS month,
  COUNT(*) AS down_days
FROM data
WHERE label = 0
  AND SUBSTR(date, 1, 4) = '2008'
GROUP BY month
ORDER BY down_days DESC
LIMIT 5
""")
```
