```python
import pandas as pd

url = "https://raw.githubusercontent.com/SujalS14/Daily-News-for-Stock-Market-Prediction_BAAI_CP/refs/heads/main/Combined_News_DJIA%202.csv"
df = pd.read_csv(url)
df.head()  #first 5 rows
```

```python
df.info()
```

```python
print(df.duplicated().sum, "duplicates")
print(df.isna().sum())

df = df.drop_duplicates()
df = df.dropna()
print(df.shape)
```

```python
df.groupby("Label").size().plot(
    kind="bar", title="Distribution of Labels"
)
```
